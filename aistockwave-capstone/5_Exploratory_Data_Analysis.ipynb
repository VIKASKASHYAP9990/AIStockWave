# Module 5: Exploratory Data Analysis & Visualization
This notebook visualizes stock sector metrics, distributions, and simulated portfolios using matplotlib and seaborn.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set design system parameters
plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)
print("Visualizers loaded.")
```

### 1. Load Data from SQLite Database

```python
conn = sqlite3.connect("aistockwave.db")
df_stocks = pd.read_sql_query("SELECT * FROM stocks", conn)
conn.close()
print(f"Loaded {len(df_stocks)} stock records for visualization.")
```

### 2. Chart 1: Top Stocks by Market Capitalization ($ Billions)
We show company value comparisons.

```python
sns.barplot(x="market_cap", y="symbol", data=df_stocks.sort_values("market_cap", ascending=False), palette="viridis")
plt.title("AIStockWave Tickers ranked by Market Capitalization ($ Billions)", fontsize=14, fontweight='bold')
plt.xlabel("Market Cap ($B)")
plt.ylabel("Ticker Ticker")
plt.tight_layout()
plt.savefig("market_cap_chart.png", dpi=150)
plt.show()
```

### 3. Chart 2: Distribution of P/E Ratios
Shows valuation frequencies.

```python
sns.histplot(df_stocks['pe_ratio'], bins=6, kde=True, color="#2962FF")
plt.title("Distribution of Price-to-Earnings (P/E) Ratios", fontsize=14, fontweight='bold')
plt.xlabel("P/E Ratio")
plt.ylabel("Frequency count")
plt.tight_layout()
plt.savefig("pe_distribution_chart.png", dpi=150)
plt.show()
```

### 4. Chart 3: Proximity channels of Live Prices against 52W Highs / Lows

```python
df_sorted = df_stocks.sort_values("current_price")
plt.scatter(df_sorted['symbol'], df_sorted['current_price'], color='blue', label='Live price', zorder=5)
plt.scatter(df_sorted['symbol'], df_sorted['high_52week'], color='green', marker='^', label='52W High', alpha=0.6)
plt.scatter(df_sorted['symbol'], df_sorted['low_52week'], color='red', marker='v', label='52W Low', alpha=0.6)
plt.title("Current Prices relative to 52-Week High & Low Boundaries", fontsize=14, fontweight='bold')
plt.ylabel("Price in ₹")
plt.legend()
plt.tight_layout()
plt.savefig('52w_channels_chart.png', dpi=150)
plt.show()
```
