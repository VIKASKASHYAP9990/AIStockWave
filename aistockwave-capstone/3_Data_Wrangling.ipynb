# Module 3: Data Cleaning and Wrangling
This notebook cleans the raw stock and news datasets, standardizes numbers, handles nulls, and exports templates ready for database insertion.

```python
import pandas as pd
import numpy as np

print("Pandas and Numpy loaded.")
```

### 1. Data Cleaning of Stock Metrics
- Check for missing values.
- Verify P/E Ratio boundaries.
- Drop duplicate symbols.

```python
df_stocks = pd.read_excel("raw_stocks_data.xlsx")

# Proactive duplicate check
duplicates = df_stocks.duplicated(subset=['Symbol']).sum()
print(f"Duplicated symbols count: {duplicates}")

# Fill empty or null values if any (e.g. PE ratio fills)
df_stocks['PE Ratio'] = df_stocks['PE Ratio'].fillna(df_stocks['PE Ratio'].mean())

# Clean symbols
df_stocks['Symbol'] = df_stocks['Symbol'].str.upper().str.strip()
print("Data cleaning checklist complete. Cleaned columns:")
print(df_stocks.dtypes)
```

### 2. Cleaning News headlines

```python
df_news = pd.read_excel("raw_news_data.xlsx")
df_news['Title'] = df_news['Title'].str.strip()
df_news['Source'] = df_news['Source'].str.strip()
print(f"Total news items cleaned: {len(df_news)}")
```

### 3. Save Cleaned Datasets to CSV files for SQL Import

```python
df_stocks.to_csv("cleaned_stocks_data.csv", index=False)
df_news.to_csv("cleaned_news_data.csv", index=False)
print("Exported cleaned_stocks_data.csv and cleaned_news_data.csv successfully.")
```
