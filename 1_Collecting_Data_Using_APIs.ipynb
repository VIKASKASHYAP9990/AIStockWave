# Module 1: Accessing Stock Market APIs
This notebook demonstrates how to access live financial APIs (e.g. Yahoo Finance, Finnhub, or Alpha Vantage) to retrieve current quotes and company statistics.

```python
import requests
import pandas as pd

print("Requests and Pandas loaded successfully.")
```

### 1. Querying Ticker Symbol Details from External REST Endpoints
We fetch details for standard companies. If API key is empty, we print a sample response and load our raw dataset spreadsheet.

```python
def fetch_stock_quote(symbol, api_key):
    if not api_key:
        print(f"No API key configured for {symbol}. Returning fallback mock response.")
        return {"symbol": symbol, "price": 185.0, "high": 186.0, "low": 183.0}
    
    endpoint = f"https://finnhub.io/api/v1/quote?symbol={symbol}&token={api_key}"
    try:
        response = requests.get(endpoint, timeout=10)
        if response.status_code == 200:
            return response.json()
    except Exception as e:
        print(f"Request failed: {e}")
    return None

quote = fetch_stock_quote("AAPL", "")
print(quote)
```
